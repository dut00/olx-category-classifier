# Libraries import

```python
import pandas as pd
import numpy as np
```

# Data preparation
## Reading files

```python
files_list = [
    "data/items.sell.csv",
    "data/items.change.csv",
    "data/items.free.csv"
]

files_key_columns = ['name', 'tag']

dfs = [pd.read_csv(file, usecols=file_key_columns) for file in files_list]

df = pd.concat(dfs, ignore_index=True)
df.tail(10)
```

## Cleaning and normalization
### Removing rows(axis=0) with missing values

```python
print(df.isna().sum())
df = df.dropna()
```

### Whitespaces normalization (replacing all whitespace characters with a single space)

```python
df["name"] = df["name"].str.replace(r"\s+", " ", regex=True).str.strip()
df["tag"] = df["tag"].str.replace(r"\s+", " ", regex=True).str.strip()
```

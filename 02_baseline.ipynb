# Baseline
## Vectorization and Classification

```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import SGDClassifier
from sklearn.pipeline import make_pipeline


train = pd.read_parquet("data/train.parquet")

clf = make_pipeline(
    TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(2, 5),
        min_df=2,
        sublinear_tf=True,
        max_features=150_000),
    # LogisticRegression(
    #     max_iter=200,
    #     C=5),             # ~1h 10min - run on 1 thread; min_df=2
    SGDClassifier(
        loss="log_loss",
        alpha=1e-5,
        max_iter=20,
        n_jobs=-1,
        random_state=42),   # 1m 49s - run on all (12) threads; min_df=3
                            # 1m 35s - run on all (12) threads; min_df=2 (wtf?)
)

clf.fit(train["name"], train["tag"])   # can take few minutes
```

### Results persistence

```python
import joblib
from pathlib import Path
import time

timestamp = round(time.time())
model_filename = f"models/baseline_tfidf_sgd_{timestamp}.joblib"

Path("models").mkdir(exist_ok=True)

# save
joblib.dump(clf, model_filename, compress=3)
```

```python
# load
clf =  joblib.load(model_filename)

# clf.predict_proba(["Sprzedam rower górski 26 cali"])
```

## Evaluation

```python
from sklearn.metrics import accuracy_score, f1_score

val = pd.read_parquet("data/val.parquet")

proba = clf.predict_proba(val["name"])
classes = clf.classes_
pred = classes[proba.argmax(axis=1)]

print("accuracy :", accuracy_score(val["tag"], pred))
print("macro F1 :", f1_score(val["tag"], pred, average="macro", zero_division=0))

top3 = classes[np.argsort(proba, axis=1)[:, -3:]]
print("top-3    :", np.mean([t in row for t, row in zip(val["tag"], top3)]))

pred_cat = pd.Series(pred).str.split(" » ").str[0].values                   # type: ignore
print("category accuracy:", (pred_cat == val["category"].values).mean())    # type: ignore
```

## Error analysis

```python
from collections import Counter
from sklearn.metrics import classification_report

val = val.assign(pred=pred)
err = val[val["tag"] != val["pred"]]
```

```python
# najczęstsze pomyłki
pairs = Counter(zip(err["tag"], err["pred"]))
for (t, p), n in pairs.most_common(15):
    print(f"{n:>4}  {t}  →  {p}")
```

```python
# najsłabsze klasy
rep = pd.DataFrame(classification_report(val["tag"], pred, output_dict=True, zero_division=0)).T
rep = rep.drop(["accuracy", "macro avg", "weighted avg"])
rep.sort_values("f1-score").head(15)
```

```python
# losowe błędne przykłady do przeczytania
err.sample(20, random_state=1)[["name", "tag", "pred"]]
```

```python
cats = pd.Series(classes).str.split(" » ").str[0].values
cat_names = np.unique(cats)
cat_proba = np.column_stack([proba[:, cats == c].sum(axis=1) for c in cat_names])
pred_cat = cat_names[cat_proba.argmax(axis=1)]
print("category accuracy (suma):", (pred_cat == val["category"].values).mean())
```
